```python
from sentence_transformers import SentenceTransformer

# Load the embedding model
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')

# Generate the sentences from previous step
sentences = "Patient from Chandler, who can speark German (Germany), belonging to ethinicity of Not Hispanic or Latino and race White, reached doctor for the service Encounter for check up (procedure) has been successfully diagnoised"

# Create embeddings
embeddings = model.encode(sentences, convert_to_tensor=True)

# Example: show shape and one vector
print(f"Total Embeddings: {embeddings.shape}")
print(f"First Vector: {embeddings[0]}")
```
