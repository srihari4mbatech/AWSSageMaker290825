```python
import boto3, awswrangler as wr

glue = boto3.client("glue", region_name="us-east-1")  # pick your region
t = glue.get_table(DatabaseName="jsondata", Name="syntheticmedicare10k")

s3_path = t["Table"]["StorageDescriptor"]["Location"]  # e.g. s3://bucket/path/
input_format = t["Table"]["StorageDescriptor"]["InputFormat"]  # to infer file type

# Example for Parquet:
print(input_format)
df = wr.s3.read_parquet(s3_path, dataset=True)  # dataset=True handles partitioned data
df.head()
```

```python
# Example for Parquet:
df = wr.s3.read_json(
    s3_path,
    dataset=True,
    lines=True,
    encoding="utf-16")  # dataset=True handles partitioned data
df.head()
```
