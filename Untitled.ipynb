```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import gradio as gr

# ------------------ Sample Patient and Doctor Data ------------------
doctors = pd.DataFrame({
    'doctor_id': [101, 102, 103],
    'name': ['Dr. Smith', 'Dr. Kumar', 'Dr. Lopez'],
    'specialty': ['Cardiologist', 'Endocrinologist', 'Pulmonologist'],
    'city': ['New York', 'San Francisco', 'Los Angeles'],
    'location': ['Manhattan', 'Bay Area', 'Downtown'],
    'preferred_duration': ['short', 'medium', 'long'],
    'languages_spoken': ['English', 'Hindi, English', 'Spanish, English'],
    'race_focus': ['White', 'Asian', 'Hispanic']
})

def merge_doctor_features(row):
    return f"{row['specialty']} {row['city']} {row['location']} {row['preferred_duration']} {row['languages_spoken']} {row['race_focus']}"

doctors['features'] = doctors.apply(merge_doctor_features, axis=1)

# ------------------ Vectorize Doctor Features ------------------
tfidf = TfidfVectorizer()
doctor_tfidf = tfidf.fit_transform(doctors['features'])

# ------------------ Define Recommendation Function ------------------
def recommend_doctor(name, symptom, city, location, encounter_duration, language, race):
    patient_features = f"{symptom} {city} {location} {encounter_duration} {language} {race}"
    patient_vector = tfidf.transform([patient_features])
    similarity_scores = cosine_similarity(patient_vector, doctor_tfidf)[0]
    top_idx = similarity_scores.argmax()
    recommended_doctor = doctors.loc[top_idx, 'name']
    return f"Patient {name} is recommended to see {recommended_doctor}" 

# ------------------ Gradio Interface ------------------
interface = gr.Interface(
    fn=recommend_doctor,
    inputs=[
        gr.Textbox(label="Patient Name"),
        gr.Textbox(label="Symptom"),
        gr.Textbox(label="City"),
        gr.Textbox(label="Location"),
        gr.Dropdown(choices=["short", "medium", "long"], label="Encounter Duration"),
        gr.Textbox(label="Language"),
        gr.Textbox(label="Race")
    ],
    outputs="text",
    title="Doctor Recommendation System",
    description="Input patient details to find the most suitable doctor."
)

interface.launch()
```

```python
interface.close()
```
