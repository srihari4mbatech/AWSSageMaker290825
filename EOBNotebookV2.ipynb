```python
import gradio as gr
from sklearn.metrics.pairwise import cosine_similarity

def recommend_doctor(name, Service, city, location, ethnicity, language, race):
    patient_features = f"{Service} {city} {location} {ethnicity} {language} {race}"
    patientrecord = f"Patient from {city}, who can speark {language}, belonging to ethinicity of {ethnicity} and race {race}, reached doctor for the service {Service} has been successfully diagnoised"
    model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
    embeddings = model.encode(patientrecord, convert_to_tensor=True)
    similarity_scores = cosine_similarity(embeddings.reshape(1,-1), embeddingVector.to_list())[0]
    top_idx = similarity_scores.argmax()
    recommended_doctor = SamplWPatientProvidedFilterD.loc[top_idx, 'Practitioner_Name']
    return f"Patient {name} is recommended to see {recommended_doctor}" 

# ------------------ Gradio Interface ------------------
interface = gr.Interface(
    fn=recommend_doctor,
    inputs=[
        gr.Textbox(label="Patient Name"),
        gr.Textbox(label="Service"),
        gr.Textbox(label="City"),
        gr.Textbox(label="Ethnicity"),
        gr.Textbox(label="Language"),
        gr.Textbox(label="Race")
    ],
    outputs="text",
    title="Doctor Recommendation System",
    description="Input patient details to find the most suitable doctor."
)

interface.launch()
```
